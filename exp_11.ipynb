```python
# Experiment 11
# Detect port-scanning behaviour from network traffic logs

from collections import defaultdict

def detect_port_scanning(log_file):

    # Store ports accessed by each IP
    connections = defaultdict(set)

    # Read network log
    with open(log_file, "r") as file:

        for line in file:

            parts = line.split()

            # Format:
            # IP_ADDRESS PROTOCOL PORT
            source_ip = parts[0]
            destination_port = int(parts[2])

            connections[source_ip].add(destination_port)

    print("PORT SCANNING DETECTION")
    print("----------------------")

    # Analyze each IP
    for ip, ports in connections.items():

        print("Source IP:", ip)
        print("Number of Ports Accessed:", len(ports))
        print("Ports:", sorted(ports))

        # If more than 5 ports are accessed,
        # consider it suspicious
        if len(ports) >= 5:

            print("Status: Possible Port Scanning Detected!")

        else:

            print("Status: Normal Network Activity")

        print()


# -----------------------------------------------
# Create sample network traffic log
# -----------------------------------------------

with open("network.log", "w") as file:

    file.write("192.168.1.100 TCP 21\n")
    file.write("192.168.1.100 TCP 22\n")
    file.write("192.168.1.100 TCP 23\n")
    file.write("192.168.1.100 TCP 25\n")
    file.write("192.168.1.100 TCP 53\n")
    file.write("192.168.1.100 TCP 80\n")
    file.write("192.168.1.100 TCP 443\n")

    file.write("192.168.1.20 TCP 80\n")
    file.write("192.168.1.20 TCP 443\n")


# Run port scan detection
detect_port_scanning("network.log")
```
