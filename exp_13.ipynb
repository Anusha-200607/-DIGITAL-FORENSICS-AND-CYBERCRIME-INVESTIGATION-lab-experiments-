```python
# Experiment 13
# Automatically Generate a Structured Digital Forensic Report

from datetime import datetime

def generate_report():

    # Get case details from the user
    case_id = input("Enter Case ID: ")
    investigator = input("Enter Investigator Name: ")
    incident = input("Enter Incident Type: ")

    # Get investigation details
    evidence = input("Enter Evidence Collected: ")
    findings = input("Enter Investigation Findings: ")
    conclusion = input("Enter Conclusion: ")

    # Create structured report
    report = f"""
========================================
       DIGITAL FORENSIC REPORT
========================================

CASE INFORMATION
----------------
Case ID       : {case_id}
Investigator  : {investigator}
Date          : {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}
Incident Type : {incident}

EVIDENCE COLLECTED
------------------
{evidence}

INVESTIGATION FINDINGS
----------------------
{findings}

CONCLUSION
----------
{conclusion}

========================================
          END OF REPORT
========================================
"""

    # Create report file
    filename = case_id + "_forensic_report.txt"

    # Save report
    with open(filename, "w") as file:
        file.write(report)

    print("\nDigital forensic report generated successfully!")
    print("Report saved as:", filename)


# Run the program
generate_report()
```
