```python
# Experiment 1.3 - Ransomware IoCs and Signatures

import re

# Simulated ransomware IoC database
ransomware_iocs = {
    "ExampleLocker": {
        "domains": [
            "examplelocker.fake",
            "payment-example.fake"
        ],
        "ip_addresses": [
            "192.0.2.10",
            "192.0.2.20"
        ],
        "file_extensions": [
            ".locked",
            ".encrypted"
        ],
        "ransom_notes": [
            "README_RECOVER.txt",
            "HOW_TO_DECRYPT.txt"
        ]
    }
}

# Simulated files
sample_files = [
    "document.docx",
    "photo.jpg.locked",
    "database.sql.encrypted",
    "README_RECOVER.txt"
]

# Simulated detection
print("=" * 60)
print("RANSOMWARE IoC AND SIGNATURE ANALYSIS")
print("=" * 60)

print("\nRansomware Family: ExampleLocker")

print("\nIoC Indicators")
print("-" * 60)

for category, values in ransomware_iocs["ExampleLocker"].items():

    print(category.upper() + ":")

    for value in values:
        print(" -", value)


print("\nFile Analysis")
print("-" * 60)

detected = False

for filename in sample_files:

    print("\nAnalyzing:", filename)

    if filename.endswith(".locked") or filename.endswith(".encrypted"):
        print("Alert: Suspicious ransomware file extension detected.")
        detected = True

    if filename in ransomware_iocs["ExampleLocker"]["ransom_notes"]:
        print("Alert: Possible ransom note detected.")
        detected = True


print("\nDetection Rule")
print("-" * 60)

print("""
RULE ExampleLocker_Detection:
IF
    file extension is .locked OR .encrypted
    OR ransom note is detected
THEN
    Generate Ransomware Alert
""")


print("\nFinal Result")
print("-" * 60)

if detected:
    print("Status: POSSIBLE RANSOMWARE ACTIVITY DETECTED")
    print("Risk Level: HIGH")
else:
    print("Status: No ransomware indicators detected.")


print("\nSimulated Ransomware Timeline")
print("-" * 60)

timeline = [
    "10:00 - Suspicious file downloaded",
    "10:02 - Multiple files renamed with .locked extension",
    "10:03 - Ransom note created",
    "10:05 - Ransomware activity detected",
    "10:06 - Security alert generated"
]

for event in timeline:
    print(event)
```
