```python
# Experiment 1.4 - Ethical Attack Pattern Detection

import re

# Simulated security logs
logs = [
    "LOGIN_FAILED user=admin ip=192.168.1.10",
    "LOGIN_FAILED user=admin ip=192.168.1.10",
    "LOGIN_FAILED user=admin ip=192.168.1.10",
    "LOGIN_FAILED user=admin ip=192.168.1.10",
    "LOGIN_FAILED user=admin ip=192.168.1.10",

    "HTTP_REQUEST path=/login?user=' OR '1'='1",

    "HTTP_REQUEST path=/../../etc/passwd",

    "LOGIN_SUCCESS user=user1 ip=192.168.1.20"
]

print("=" * 60)
print("ETHICAL ATTACK PATTERN DETECTION")
print("=" * 60)

# Counters
failed_logins = {}
sql_injection_count = 0
directory_traversal_count = 0

print("\nAnalyzing Security Logs")
print("-" * 60)

for log in logs:

    print("\nLog:", log)

    # Detect failed login
    if "LOGIN_FAILED" in log:

        ip_match = re.search(r"ip=([0-9.]+)", log)

        if ip_match:

            ip = ip_match.group(1)

            failed_logins[ip] = failed_logins.get(ip, 0) + 1

    # Detect SQL injection patterns
    sql_patterns = [
        r"' OR '1'='1",
        r"UNION SELECT",
        r"DROP TABLE",
        r"--"
    ]

    for pattern in sql_patterns:

        if re.search(pattern, log, re.IGNORECASE):

            sql_injection_count += 1
            print("ALERT: Possible SQL Injection Pattern Detected")
            break

    # Detect directory traversal
    if "../" in log or "..%2f" in log.lower():

        directory_traversal_count += 1
        print("ALERT: Possible Directory Traversal Detected")


# Brute-force detection
print("\nBrute-Force Analysis")
print("-" * 60)

for ip, count in failed_logins.items():

    print("IP Address:", ip)
    print("Failed Login Attempts:", count)

    if count >= 5:
        print("ALERT: Possible Brute-Force Attack")


# Final report
print("\n" + "=" * 60)
print("VULNERABILITY CLASSIFICATION REPORT")
print("=" * 60)

print("\nBrute-Force Attempts:", sum(failed_logins.values()))
print("SQL Injection Attempts:", sql_injection_count)
print("Directory Traversal Attempts:", directory_traversal_count)

print("\nWAF RULE SUGGESTIONS")
print("-" * 60)

print("1. Rate-limit repeated failed login attempts.")
print("2. Detect and block suspicious SQL injection patterns.")
print("3. Block directory traversal sequences such as ../.")
print("4. Log and monitor suspicious HTTP requests.")

print("\nFinal Status")
print("-" * 60)

if sum(failed_logins.values()) >= 5:
    print("Brute-Force: DETECTED")

if sql_injection_count > 0:
    print("SQL Injection: DETECTED")

if directory_traversal_count > 0:
    print("Directory Traversal: DETECTED")
```
