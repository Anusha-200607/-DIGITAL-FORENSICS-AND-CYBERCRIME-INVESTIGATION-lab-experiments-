```python
# Digital Evidence Collection & Cataloguing

evidence_catalog = []

def add_evidence(evidence_id, evidence_type, description, source, collector):
    evidence = {
        "Evidence ID": evidence_id,
        "Type": evidence_type,
        "Description": description,
        "Source": source,
        "Collector": collector
    }
    evidence_catalog.append(evidence)

add_evidence(
    "E001",
    "Hard Disk",
    "Suspect computer hard disk",
    "Computer System 01",
    "Forensic Investigator"
)

add_evidence(
    "E002",
    "USB Drive",
    "USB storage device",
    "Evidence Room",
    "Forensic Investigator"
)

add_evidence(
    "E003",
    "Mobile Phone",
    "Smartphone collected from scene",
    "Crime Scene",
    "Forensic Investigator"
)

print("DIGITAL EVIDENCE CATALOG")
print("-" * 50)

for evidence in evidence_catalog:
    for key, value in evidence.items():
        print(f"{key}: {value}")
    print("-" * 50)
```
