```python
# Volatile vs Non-Volatile Evidence

volatile_evidence = [
    "RAM contents",
    "Running processes",
    "Network connections",
    "CPU registers",
    "System uptime"
]

non_volatile_evidence = [
    "Hard disk files",
    "SSD data",
    "USB drive data",
    "Log files",
    "Deleted files"
]

print("VOLATILE EVIDENCE")
print("=" * 30)

for item in volatile_evidence:
    print("-", item)

print("\nNON-VOLATILE EVIDENCE")
print("=" * 30)

for item in non_volatile_evidence:
    print("-", item)

print("\nRecommended Capture Order:")
print("1. Capture volatile evidence first.")
print("2. Capture non-volatile evidence after securing the system.")
```
