```python
# Physical, Logical and Sparse Acquisition Simulation

data_blocks = [
    "Block 1: User Document",
    "Block 2: Empty",
    "Block 3: Image File",
    "Block 4: Empty",
    "Block 5: System File",
    "Block 6: Empty"
]

print("ORIGINAL STORAGE")
print("=" * 40)

for block in data_blocks:
    print(block)

print("\nPHYSICAL ACQUISITION")
print("=" * 40)

for block in data_blocks:
    print(block)

print("\nLOGICAL ACQUISITION")
print("=" * 40)

for block in data_blocks:
    if "Empty" not in block:
        print(block)

print("\nSPARSE ACQUISITION")
print("=" * 40)

for block in data_blocks:
    if "Empty" not in block:
        print(block)
```
