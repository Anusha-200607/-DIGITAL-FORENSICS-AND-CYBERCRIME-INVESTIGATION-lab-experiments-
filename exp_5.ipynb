```python
from datetime import datetime

def generate_forensic_report():
    report = f"""
========================================
 DIGITAL FORENSIC INVESTIGATION REPORT
========================================
Date: {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}

1. IDENTIFICATION
------------------
Identify the incident, affected systems, devices, and potential sources
of digital evidence.

2. PRESERVATION
----------------
Preserve digital evidence to prevent modification or contamination.
Maintain evidence integrity using hashing and proper documentation.

3. COLLECTION
--------------
Collect relevant digital evidence such as files, logs, emails,
network data, and storage media.

4. EXAMINATION AND ANALYSIS
----------------------------
Examine the collected evidence and analyze it to identify relevant
information, suspicious activities, and possible attack patterns.

5. REPORTING
-------------
Document the investigation process, evidence collected, findings,
analysis results, and final conclusions.

========================================
 END OF FORENSIC REPORT
========================================
"""

    with open("forensic_lifecycle_report.txt", "w") as file:
        file.write(report)

    print("Forensic report generated successfully.")

generate_forensic_report()
```
