```python
import hashlib
import shutil

# Step 1: Create a sample evidence file
source = "evidence.txt"

with open(source, "w") as file:
    file.write("This is sample digital forensic evidence.")

# Step 2: Name of forensic copy
forensic_copy = "evidence_copy.dd"

# Step 3: Create a bit-stream copy
shutil.copyfile(source, forensic_copy)

# Step 4: Function to calculate SHA-256 hash
def calculate_hash(filename):
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            chunk = file.read(4096)

            if not chunk:
                break

            sha256.update(chunk)

    return sha256.hexdigest()

# Step 5: Calculate hashes
original_hash = calculate_hash(source)
copy_hash = calculate_hash(forensic_copy)

# Step 6: Display hashes
print("Original File Hash:")
print(original_hash)

print("\nForensic Copy Hash:")
print(copy_hash)

# Step 7: Verify integrity
if original_hash == copy_hash:
    print("\nIntegrity Verified!")
    print("The forensic copy is identical to the original file.")
else:
    print("\nIntegrity Verification Failed!")
    print("The files are different.")
```
