```python
# Program to identify a file's true type using Magic Numbers

def identify_file_type(filename):
    # Read the first 8 bytes of the file
    with open(filename, "rb") as file:
        signature = file.read(8)

    # Common file signatures (Magic Numbers)
    signatures = {
        b"\x89PNG\r\n\x1a\n": "PNG Image",
        b"\xff\xd8\xff": "JPEG Image",
        b"GIF87a": "GIF Image",
        b"GIF89a": "GIF Image",
        b"%PDF": "PDF Document",
        b"PK\x03\x04": "ZIP Archive / Office Document",
        b"MZ": "Windows Executable"
    }

    # Check the signature
    for magic, file_type in signatures.items():
        if signature.startswith(magic):
            return file_type

    return "Unknown File Type"


# Step 1: Create a sample file with a PNG magic number
filename = "sample_file"

with open(filename, "wb") as file:
    file.write(b"\x89PNG\r\n\x1a\n")
    file.write(b"Sample file data")


# Step 2: Identify the true file type
print("File:", filename)
print("True File Type:", identify_file_type(filename))
```
