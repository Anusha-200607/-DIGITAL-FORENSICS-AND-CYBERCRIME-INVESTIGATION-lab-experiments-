```python
# Experiment 8
# Simulate recovery of deleted files using recognizable byte patterns

def recover_files(image_file):
    # File signatures and their ending markers
    signatures = {
        b"\xff\xd8\xff": ("jpg", b"\xff\xd9"),       # JPEG
        b"\x89PNG\r\n\x1a\n": ("png", b"IEND"),       # PNG
        b"%PDF": ("pdf", b"%%EOF")                    # PDF
    }

    # Read the binary storage image
    with open(image_file, "rb") as file:
        data = file.read()

    recovered_count = 0

    # Search for each file type
    for signature, (extension, footer) in signatures.items():
        start = 0

        while True:
            # Find the file header
            start = data.find(signature, start)

            if start == -1:
                break

            # Find the file footer
            end = data.find(footer, start)

            if end != -1:
                end += len(footer)

                # Extract the file data
                recovered_data = data[start:end]

                # Create recovered file
                filename = f"recovered_{recovered_count}.{extension}"

                with open(filename, "wb") as output:
                    output.write(recovered_data)

                print("Recovered file:", filename)

                recovered_count += 1

            start += len(signature)

    print("\nTotal files recovered:", recovered_count)


# ------------------------------------------------
# Create a simulated storage image
# ------------------------------------------------

with open("disk_image.bin", "wb") as file:
    # Random data
    file.write(b"RANDOM DATA 12345")

    # Simulated JPEG file
    file.write(b"\xff\xd8\xff")
    file.write(b"This is a deleted JPEG file")
    file.write(b"\xff\xd9")

    # More random data
    file.write(b"RANDOM DATA")

    # Simulated PDF file
    file.write(b"%PDF")
    file.write(b"This is a deleted PDF file")
    file.write(b"%%EOF")


# Run file recovery
recover_files("disk_image.bin")
```
