```python
# Experiment 9
# Detect brute-force intrusion attempts from a login log

from collections import defaultdict

def detect_brute_force(log_file):

    # Dictionary to store failed attempts
    failed_attempts = defaultdict(int)

    # Read the log file
    with open(log_file, "r") as file:

        for line in file:

            if "Failed login" in line:

                # Get IP address
                parts = line.split()
                ip_address = parts[-1]

                # Increase failed attempt count
                failed_attempts[ip_address] += 1

    print("BRUTE-FORCE ATTACK DETECTION")
    print("----------------------------")

    # Check each IP address
    for ip, count in failed_attempts.items():

        print("IP Address:", ip)
        print("Failed Attempts:", count)

        if count >= 5:
            print("Status: Possible Brute-Force Attack!")
        else:
            print("Status: Normal")

        print()


# -----------------------------------------------
# Create a sample login log
# -----------------------------------------------

with open("login.log", "w") as file:

    file.write("Failed login user1 192.168.1.10\n")
    file.write("Failed login user1 192.168.1.10\n")
    file.write("Failed login user1 192.168.1.10\n")
    file.write("Failed login user1 192.168.1.10\n")
    file.write("Failed login user1 192.168.1.10\n")

    file.write("Successful login user2 192.168.1.20\n")


# Run brute-force detection
detect_brute_force("login.log")
```
